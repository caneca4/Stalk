```python
import pandas as pd
from datetime import datetime
from IPython.display import display, clear_output, HTML
import ipywidgets as widgets

# Base de dados em memória
usuarios = {}
tweets = []

# Usuário logado (usamos lista para manter referência)
usuario_logado = [None]
```

```python
# Login e cadastro
login_input = widgets.Text(placeholder='Usuário', description='Usuário:')
senha_input = widgets.Password(placeholder='Senha', description='Senha:')
botao_login = widgets.Button(description='Entrar')
botao_cadastro = widgets.Button(description='Cadastrar')

# Postagem de tweet
tweet_input = widgets.Textarea(placeholder='Escreva seu post...', description='Post:')
botao_postar = widgets.Button(description='Postar')

# Áreas de saída
saida_feed = widgets.Output()
saida_status = widgets.Output()
```

```python
# Função para mostrar a tela de login
def mostrar_tela_login():
    clear_output()
    display(HTML("<h2>Bem-vindo ao Stalk!</h2>"))
    display(login_input, senha_input, widgets.HBox([botao_login, botao_cadastro]), saida_status)

# Função para mostrar o feed de tweets
def mostrar_feed():
    clear_output()
    display(HTML(f"<h3>Olá, {usuario_logado[0]}</h3>"))
    display(tweet_input, botao_postar)
    display(widgets.HTML("<h4>Feed:</h4>"))
    mostrar_tweets()

# Função para mostrar todos os tweets
def mostrar_tweets():
    saida_feed.clear_output()
    with saida_feed:
        if tweets:
            for tweet in reversed(tweets):
                print(f"{tweet['autor']} - {tweet['data']}")
                print(tweet['texto'])
                print('-' * 40)
        else:
            print("Nenhum post ainda.")
    display(saida_feed)
```

```python
# Função de login
def login_usuario(b):
    user = login_input.value.strip()
    senha = senha_input.value.strip()
    if user in usuarios and usuarios[user] == senha:
        usuario_logado[0] = user
        mostrar_feed()
    else:
        saida_status.clear_output()
        with saida_status:
            print("Usuário ou senha inválidos.")

# Função de cadastro de usuário
def cadastrar_usuario(b):
    user = login_input.value.strip()
    senha = senha_input.value.strip()
    if user in usuarios:
        saida_status.clear_output()
        with saida_status:
            print("Usuário já existe.")
    elif not user or not senha:
        saida_status.clear_output()
        with saida_status:
            print("Preencha usuário e senha.")
    else:
        usuarios[user] = senha
        saida_status.clear_output()
        with saida_status:
            print("Cadastro realizado. Agora faça login.")

# Função de postagem de tweet
def postar(b):
    texto = tweet_input.value.strip()
    if texto:
        tweets.append({
            'autor': usuario_logado[0],
            'texto': texto,
            'data': datetime.now().strftime('%d/%m/%Y %H:%M:%S')
        })
        tweet_input.value = ''
        mostrar_feed()
```

```python
# Eventos de cliques
botao_login.on_click(login_usuario)
botao_cadastro.on_click(cadastrar_usuario)
botao_postar.on_click(postar)

# Inicia a tela de login
mostrar_tela_login()
```
